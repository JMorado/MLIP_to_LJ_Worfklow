```python
import pandas as pd
import os
import numpy as np
```

```python
df = pd.read_csv('../training-set-renamed-with-mapped-smiles.csv')
```

```python
ids = df["Id"].tolist()
mols1 = df["Component1-Mapped"].tolist() 
mols2 = df["Component2-Mapped"].tolist() 
x1s = df["Mole Fraction 1"].tolist() 
x2s = df["Mole Fraction 2"].tolist()
densities = df['Density Value (g / ml)'].tolist()
```

```python
PREFIX = "mixtures_training-set"
count = 0
for id, mol1, mol2, x1, x2, dens in zip(ids, mols1, mols2, x1s, x2s, densities):
    os.makedirs(f"{PREFIX}/run_{count:04d}", exist_ok=True)
    template = open("template_mapped.py").read()
    mol1 = mol1 if mol1 is not np.nan else ''
    mol2 = mol2 if mol2 is not np.nan else ''
    if mol2 == '':
        x2 = 0.0
    density = dens if dens is not np.nan else 1.0

    with open(f"{PREFIX}/run_{count:04d}/run_{id}.py", "w") as f:
        script = template.replace("__SMILES1__", f'"{mol1}"')
        script = script.replace("__SMILES2__", f'"{mol2}"')
        script = script.replace("__X1__", str(x1))
        script = script.replace("__X2__", str(x2))
        script = script.replace("__DENSITY__", f"{density}")
        f.write(script)
    count += 1
```
